```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Histrograms  = A histogram is a graph that shows how numerical observations 
#                are distributed across intervals.
```

```python
scores = np.random.normal(loc= 80 ,scale =10 , size=100) #loc  - > all of the numbers will cluster arround 80  
                                              #scale -> how far away the numbers deviate from the center  
                                              #size => how many numbers to generate 

scores = np.clip(scores,0,100) #any score below 0 will be clipped to 0
                               #any score above 100 will be cliped to 100
plt.hist(scores, bins = 10 ,# bins = number of columns like structures
         color = "green",
         edgecolor="black"
        ) 
plt.title("Scores")
plt.xlabel("Score")
plt.ylabel("number of Students")
plt.show()
```
