```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2021,2022,2023,2024])
y = np.array([7,43,66,1])
z = np.array([11,30,14,19])
```

```python
l_st = dict(marker="v" ,  # o v * 1 2 3 4 5  
             markersize=1, #ms -> marker size
             markerfacecolor='green',
             linestyle="dashdot", # dashdot
             linewidth=2,
             color="orange")
```

```python
plt.plot(x,y,**l_st)
plt.plot(x,z,**l_st)

plt.title("Class size",
          fontsize=20,
          family="Arial",
          fontweight='bold',
          color="pink"
         )
plt.xlabel("Year",fontsize=12)
plt.ylabel("students ")
plt.tick_params(axis='both',color="green")
plt.xticks(x)
plt.subplots_adjust(bottom=0.145 ,left=0.14)
plt.show()
```
