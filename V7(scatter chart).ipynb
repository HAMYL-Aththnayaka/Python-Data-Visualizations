```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Scatter graph = shows the  relationship between 2 variables 
#                Helps to identify a correlation ( +,-,None )
#                Example : Study hours vs Test scores
```

```python
#class A
x1 = np.array([ 0 , 1 , 1, 2, 3, 4, 5, 6, 7, 7, 8]) #hours studies
y1 = np.array([55 , 60 , 65 , 62 , 68 , 70 ,75 , 78 , 82 , 85 , 87]) # Grades

#class B
x2 = np.array([ 0 , 1 ,2, 2, 3, 4, 5, 6, 8, 7, 8]) #hours studies
y2 = np.array([50 , 78 , 65 , 43 , 18 , 76 ,76 , 98 , 52 , 45 , 37]) # Grades
```

```python
#plt.scatter(x , y , color ="cyan" ,
#            alpha =0.4 ,#Transparency
#            s = 10 #size
#           )


plt.scatter(x1 , y1 , color ="cyan" ,
            alpha =0.4 ,#Transparency
            s = 10 ,#size
            label = "class A"
           )
plt.scatter(x2 , y2 , color ="red" ,
            alpha =0.4 ,#Transparency
            s = 10 ,#size
            label = "class B"   
           )

plt.title("Test Scores")
plt.xlabel("Hours Studies")
plt.ylabel("Grades")

plt.legend()
plt.show()
```
