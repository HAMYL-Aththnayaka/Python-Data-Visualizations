```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# bar chart = Circular chart divided into slice to show percentages of the total
#           # Good for visualizing distributing among categories
```

```python
catagoris = ["Freshmen" ,"Sophomores" ,"Juniors" ,"Seniors" ]
values = np.array([300,250,275,225])
```

```python
colors = ["red","yellow","blue","green"]
```

```python
plt.pie(values ,
        labels=catagoris ,
        autopct="%1.1f%%" ,
        colors =colors ,
        explode=[0,0,0,0.09] ,
        shadow = True ,
        startangle=180
       )
plt.title("Yzz Collage")
plt.show()
```
