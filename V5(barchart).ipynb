```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#bar chart = compare catergories of data by representing each catagory with a bar
```

```python
catagories = ["Grains" , "Fruit" , "Vegetable" ,"Protein","Dairy","Sweets"]
values = np.array([4,3,2,5,3,1])
```

```python
plt.bar(catagories , values)
plt.title("Dairy Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```
