```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2021,2022,2023,2024])
y = np.array([7,43,66,1])
z = np.array([11,30,14,19])
```

```python
plt.plot(x,y,marker="o" ,  # o v * 1 2 3 4 5  
             markersize=10, #ms -> marker size
             markerfacecolor='cyan',
             linestyle="solid", # dashdot
             linewidth=2,
             color="purple")
l_st = dict(marker="v" ,  # o v * 1 2 3 4 5  
             markersize=1, #ms -> marker size
             markerfacecolor='green',
             linestyle="dashdot", # dashdot
             linewidth=2,
             color="orange")
plt.plot(x,z,**l_st)
plt.show()
```
